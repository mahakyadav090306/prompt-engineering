```
import google.generativeai as genai

genai.configure(api_key="AQ.Ab8RN6KWtgO9QJyEbWOSkeUtDxt3OssrHoG0J8v2mdTrkeRvCw")

model = genai.GenerativeModel("models/gemini-2.5-flash")
```

```
prompt = '''
You are an expert data extraction system.

Extract the required information from the given unstructured text.

Rules:
- The text will always be inside triple quotes (""").
- Return ONLY valid JSON.
- No explanation.
- No markdown.
- Missing values should be null.

Output format:

{
"name":"",
"email":"",
"phone":"",
"city":"",
"skills":[],
"experience_years":null
}

Example 1

Input:

"""
John Smith lives in Delhi.
Email: john@gmail.com
Phone: 9876543210
Skills: Python, SQL
Experience: 3 years
"""

Output:

{
"name":"John Smith",
"email":"john@gmail.com",
"phone":"9876543210",
"city":"Delhi",
"skills":["Python","SQL"],
"experience_years":3
}

Example 2

Input:

"""
Emily Brown
Lives in Mumbai.
Email: emily@gmail.com
Skills: Java, C++
Experience: 5 years
"""

Output:

{
"name":"Emily Brown",
"email":"emily@gmail.com",
"phone":null,
"city":"Mumbai",
"skills":["Java","C++"],
"experience_years":5
}

Now extract:

"""
My name is Mahak Yadav.

I live in Rewari.

Email:
mahak@gmail.com

Phone:
9876543211

Skills:
Python
Java
Machine Learning
SQL

Experience:
2 years
"""
'''
```

```
response = model.generate_content(
    prompt,
    generation_config={
        "temperature": 0
    }
)

print(response.text)
```
