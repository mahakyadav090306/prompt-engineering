```
model = genai.GenerativeModel("models/gemini-2.5-flash")
```

```
prompt = '''
You are a Logic Reasoning Engine.

Your task is to solve logic problems carefully.

Follow these rules:

1. Read the problem carefully.
2. Break the solution into numbered reasoning steps.
3. Verify every important assumption.
4. Perform a self-check before giving the answer.
5. If an inconsistency is found, correct it before producing the final answer.
6. Finally provide the answer in the following format ONLY:

Problem:
...

Reasoning Summary:
1.
2.
3.

Self-Check:
- Verified assumptions
- Checked calculations
- Looked for logical errors

Final Answer:
...

Problem:

A farmer has 17 sheep.
All but 9 die.

How many sheep are left?
'''
```

```
response = model.generate_content(
    prompt,
    generation_config={
        "temperature": 0
    }
)

print(response.text)
```
